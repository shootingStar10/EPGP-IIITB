# Lesk Algorithm to detect Word Sense Disambiguation

```python
# import libraries
import nltk
from nltk.corpus import wordnet
from nltk import wsd
import spacy
import warnings
```

```python
nltk.download('wordnet')
```

```python
# define sentences
X = 'The die is cast.'
Y = 'Roll the die to get a 6.'
Z = 'What is dead may never die.'
```

```python
# all synsets of word die
wordnet.synsets('die')
```

```python
# noun synsets of word die
wordnet.synsets('die', pos = wordnet.NOUN)
```

```python
# all noun senses definition of word die
for syn in wordnet.synsets('die', pos=wordnet.NOUN):
    print(syn.definition())
```

```python
# all verb senses definition of word die
for syn in wordnet.synsets('die', pos=wordnet.VERB):
    print(syn.definition())
```

```python
# pass without pos
print(X)
print(wsd.lesk(X, 'die').definition())

print()

# pass with pos
print(wsd.lesk(X, 'die', pos = wordnet.NOUN).definition())
```

```python
# pass without pos
print(Y)
print(wsd.lesk(Y, 'die').definition())

print()

# pass with pos
print(wsd.lesk(Y, 'die', pos = wordnet.NOUN).definition())
```

```python
# pass without pos
print(Z)
print(wsd.lesk(Z, 'die').definition())

print()

# pass with pos
print(wsd.lesk(Z, 'die', pos = wordnet.VERB).definition())
```

```python
# define lesk function without manually pos passing

POS_MAP = {
    'VERB': wordnet.VERB,
    'NOUN': wordnet.NOUN,
    'PROPN': wordnet.NOUN
}

nlp = spacy.load('en_core_web_sm')

def lesk(sentence, word):

  tokens = nlp(sentence)
  found = False
  token = None

  for tok in tokens:
    if tok.text == word:
      token = tok
      found = True
      break

  if found == False:
    raise ValueError(f'Word \"{word}\" does not appear in the sentence: {sentence}.')

  pos = POS_MAP.get(token.pos_, None)

  if pos == None:
    warnings.warn(f'POS tag for {word.text} not found in wordnet. Falling back to default Lesk behaviour.')

  return wsd.lesk(sentence, word, pos=pos)
```

```python
# using lesk, sense of die in X
print(X)
lesk(X, 'die').definition()
```

```python
# using lesk, sense of die in Y
print(Y)
lesk(Y, 'die').definition()
```

```python
# using lesk, sense of die in Z
print(Z)
lesk(Z, 'die').definition()
```

```python
def cosine_similarity(x1, y1, x2, y2):
  dot = (x1*x2) + (y1*y2)
  norm1 = (x1**2 + y1**2)**0.5
  norm2 = (x2**2 + y2**2)**0.5
  return round(dot/(norm1*norm2), 3)
```

```python
cosine_similarity(3, 8, 4, 6)
```

```python
cosine_similarity(4, 6, -12, 8)
```
