# Text pre-processing with Keras

```python
# import libraries
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D
import numpy as np
```

## Tokenization

```python
# tokenize setences
sentences = ['The quick brown fox jumps over the lazy dog.']
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
```

```python
# word index
tokenizer.word_index
```

```python
# build train sentences
train_sequences = tokenizer.texts_to_sequences(sentences)
train_sequences = np.array(train_sequences)
print(train_sequences)
```

## Creating embedding layer

```python
# define embedding
embedding = Embedding(input_dim = len(train_sequences[0]), output_dim = 128)
```

```python
# create train sample using embedding
train_sample = embedding(train_sequences)
```

```python
# print shape
print(train_sequences.shape)
print(train_sample.shape)
```

```python
# view train_sample entry
train_sample[0]
```

## Average across tokens

```python
GlobalAveragePooling1D()(train_sample)
```

## Creating Word Embeddings for more than one sentence

```python
# more than one sentence
test_corpus = [
    'The quick brown fox jumps over the lazy dog.',
    'The quick brown fox.',
    'The lazy dog.',
    'The dog.',
    'Dog and the fox.',
    'Hello, world!'
]

encoded_sentences = tokenizer.texts_to_sequences(test_corpus)
for sentence, encoded_sentence in zip(test_corpus, encoded_sentences):
    print(sentence, encoded_sentence)
```

## Padding sentences

As lengths of corpus sentences are not equal so we need to add padding.

```python
# compute max length
max_sequence_len = max([len(encoded_sentence) for encoded_sentence in encoded_sentences])
max_sequence_len
```

```python
X = pad_sequences(encoded_sentences, maxlen=max_sequence_len)
X
```

## Add embedding layer

```python
X.shape
```

```python
# X embedding computation
X_em = embedding(X)
```

```python
X_em.shape
```

```python
X_em
```

## Average across tokens

```python
z = GlobalAveragePooling1D()(X_em)
z.shape
```
